```python
import pyodbc
import json

def export_customers_to_json(server, database, username, password, output_file):
    """
    Kết nối SQL Server và export dữ liệu từ bảng Customer ra file JSON
    
    Tham số:
        server (str): Tên server SQL
        database (str): Tên database
        username (str): Tên đăng nhập SQL
        password (str): Mật khẩu
        output_file (str): Đường dẫn file output JSON
    """
    connection_string = f"""
        DRIVER={{ODBC Driver 17 for SQL Server}};
        SERVER={server};
        DATABASE={database};
        UID={username};
        PWD={password};
    """
    
    try:
        # Kết nối database
        conn = pyodbc.connect(connection_string)
        cursor = conn.cursor()
        
        # Thực hiện truy vấn
        cursor.execute("select CIF_NO,BR_CODE ,FIRST_NAME,MIDDLE_NAME,LAST_NAME,CIF_STATUS,ID_TYPE,ID_NUMBER,ID_ISSUE_PLACE,ID_ISSUE_DT,ID_EXPIRY_DT,ID_STATUS,ADDRESS_TYPE,ADDRESS,CITY,STATE,COUNTRY,POSTAGE_CODE,PHONE_NO from CIF01")       
        # Lấy tên cột
        columns = [column[0] for column in cursor.description]
        
        # Tạo danh sách dictionary chứa dữ liệu
        customers = []
        for row in cursor.fetchall():
            customers.append(dict(zip(columns, row)))
        
        # Xuất ra file JSON
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(customers, f, ensure_ascii=False, indent=4)
            
        print(f"Đã export thành công {len(customers)} khách hàng ra file {output_file}")
    
    except Exception as e:
        print(f"Lỗi: {str(e)}")
    
    finally:
        # Đảm bảo đóng kết nối
        if 'conn' in locals():
            conn.close()
```

```python
def main():
    # Thông tin mặc định
    export_customers_to_json(
        server="localhost",
        database="ISEM_PRD",
        username="sa",
        password="Tess_123456",
        output_file="customers.json"
    )

if __name__ == "__main__":
    main()
```

```python
from reportlab.lib.pagesizes import letter,A4
from reportlab.pdfgen import canvas
from reportlab.lib import colors
from reportlab.platypus import SimpleDocTemplate, Paragraph, Table, Spacer, Image
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.pdfbase import pdfmetrics
from reportlab.pdfbase.ttfonts import TTFont
```

```python

# Lấy style và áp dụng font Arial
pdfmetrics.registerFont(TTFont("Arial", r"C:\Windows\Fonts\Arial.ttf"))
pdfmetrics.registerFont(TTFont("Arial-Bold", r"C:\Windows\Fonts\arialbd.ttf"))
styles = getSampleStyleSheet()
style = styles["Normal"]
style.fontName = "Arial"  # Sử dụng font vừa đăng ký
style.fontSize = 12

# Tạo file PDF
pdf_file = "sample_report.pdf"
doc = SimpleDocTemplate(pdf_file, pagesize=A4)

# Danh sách các phần tử sẽ thêm vào PDF
elements = []

# 1. Thêm tiêu đề
styles = getSampleStyleSheet()
style_heading = styles["Heading1"]  # Lấy style Heading1
style_heading.fontName = "Arial-Bold"    # Gán font Arial để hỗ trợ tiếng Việt
style_heading.fontSize = 16         # Tùy chọn: điều chỉnh cỡ chữ nếu cần
style_heading.alignment = 2
title = Paragraph("<strong>Báo cáo mẫu</strong>", style_heading)


# 4. Thêm hình ảnh (nếu có)
try:
    img = Image(r"C:\Users\Hoang\Pictures\logo-ngan-hang-TPbank.png", width=150, height=100,hAlign='CENTER')  # Thay "example.jpg" bằng đường dẫn thực tế

except:
    elements.append(Paragraph("Không tìm thấy hình ảnh.", styles["Normal"]))
    
data = [[title, img]]
table_1 = Table(data, colWidths=[400, 150])  # Cột 1: 400pt cho title, Cột 2: 150pt cho hình
table_1.setStyle([
    ('VALIGN', (0, 0), (-1, -1), 'MIDDLE'),  # Căn giữa dọc
    ('ALIGN', (1, 0), (1, 0), 'RIGHT'),      # Căn hình sang phải
])

# Thêm bảng vào elements
elements.append(table_1)
elements.append(Spacer(1, 12))


# 2. Thêm đoạn văn
text = "Đây là một báo cáo mẫu được tạo bằng ReportLab. Nó bao gồm tiêu đề, đoạn văn, bảng và hình ảnh."
paragraph = Paragraph(text, style)
elements.append(paragraph)
elements.append(Spacer(1, 12))


# 3. Thêm bảng
data = [
    ["Tên", "Tuổi", "Địa chỉ"],
    ["Nguyễn Văn A", "25", "Hà Nội"],
    ["Trần Thị B", "30", "TP.HCM"]
]
table = Table(data)
table.setStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.grey),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'CENTER'),
    ('GRID', (0, 0), (-1, -1), 1, colors.black),
    ('FONTNAME', (0, 0), (-1, -1), "Arial")
])
elements.append(table)
elements.append(Spacer(1, 12))



# 5. Tạo PDF
doc.build(elements)

print(f"PDF đã được tạo: {pdf_file}")
```

```python
import requests
import xml.etree.ElementTree as ET
```

```python
server_url = "http://10.0.0.47:8080/jasperserver"
username = "jasperadmin"
password = "jasperadmin"

try:
    response = requests.get(
        f"{server_url}/rest_v2/serverInfo",
        auth=(username, password)
    )
    print(f"Status Code: {response.status_code}")
    print(f"Response Text: {response.text}")
    
    if response.status_code == 200:
        try:
            # Phân tích XML
            root = ET.fromstring(response.text)
            server_info = {
                "version": root.find("version").text,
                "edition": root.find("edition").text,
                "build": root.find("build").text,
                "dateFormatPattern": root.find("dateFormatPattern").text,
                "datetimeFormatPattern": root.find("datetimeFormatPattern").text
            }
            print("REST API hoạt động:", server_info)
        except ET.ParseError as e:
            print(f"Lỗi phân tích XML: {e}")
            print("Nội dung phản hồi không phải XML hợp lệ.")
    else:
        print(f"Lỗi: {response.status_code} - {response.text}")
except requests.exceptions.RequestException as e:
    print(f"Lỗi kết nối: {e}")
```
