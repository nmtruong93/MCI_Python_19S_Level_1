```python
from pymongo import MongoClient
from datetime import datetime
import pandas as pd
import json
```

```python
def aggregate_data(connection_string, database_name, collection_name, pipeline):
    try:
        # Kết nối đến MongoDB
        client = MongoClient(connection_string)
        db = client[database_name]
        collection = db[collection_name]

        # Thực hiện aggregation
        result = collection.aggregate(pipeline)

        # Trả về kết quả dưới dạng danh sách
        return list(result)
    
    except Exception as e:
        print(f"An error occurred: {e}")
        return []
```

```python
def generate_json_string(input_list):
    # Tạo dictionary từ input_list với key là giá trị bỏ ký tự '$' của phần tử
    result = {
        item[1:]: item for item in input_list
    }
    return json.dumps(result)
```

```python
def find_documents_by_field(collection, field_name, value):
    """
    Tìm các tài liệu trong MongoDB dựa trên giá trị của một trường cụ thể.
    
    Args:
        collection (pymongo.collection.Collection): Collection cần truy vấn.
        field_name (str): Tên trường cần tìm.
        value (Any): Giá trị của trường cần tìm.

    Returns:
        list: Danh sách các tài liệu phù hợp.
    """
    query = {field_name: value}
    return list(collection.find(query))
```
