```python
class MongoDBHandler:
    def __init__(self, uri: str, database_name: str):
        """
        Initialize the MongoDBHandler.
        
        :param uri: MongoDB connection URI.
        :param database_name: Name of the database to connect to.
        """
        try:
            self.client = MongoClient(uri)
            self.database = self.client[database_name]
            print(f"Connected to MongoDB database: {database_name}")
        except ConnectionFailure as e:
            print(f"Failed to connect to MongoDB: {e}")
            raise

    def insert_document(self, collection_name: str, document: dict):
        """
        Insert a single document into a collection.
        
        :param collection_name: Name of the collection.
        :param document: The document to insert (as a dictionary).
        :return: The inserted document's ID.
        """
        collection = self.database[collection_name]
        result = collection.insert_one(document)
        print(f"Inserted document ID: {result.inserted_id}")
        return result.inserted_id

    def update_document(self, collection_name: str, filter_query: dict, update_values: dict):
        """
        Update documents in a collection based on a filter.
        
        :param collection_name: Name of the collection.
        :param filter_query: The filter query to match documents.
        :param update_values: The values to update (e.g., {"$set": {...}}).
        :return: The number of documents updated.
        """
        collection = self.database[collection_name]
        result = collection.update_many(filter_query, update_values)
        print(f"Matched {result.matched_count} documents, updated {result.modified_count} documents.")
        return result.modified_count

    def delete_document(self, collection_name: str, filter_query: dict):
        """
        Delete documents from a collection based on a filter.
        
        :param collection_name: Name of the collection.
        :param filter_query: The filter query to match documents.
        :return: The number of documents deleted.
        """
        collection = self.database[collection_name]
        result = collection.delete_many(filter_query)
        print(f"Deleted {result.deleted_count} documents.")
        return result.deleted_count

    def close_connection(self):
        """
        Close the connection to MongoDB.
        """
        self.client.close()
        print("Closed connection to MongoDB.")
```

```python
class MongoDBConfig:
    # Thuộc tính tĩnh để lưu thông tin kết nối
    host = "localhost"
    port = 27017
    database = "local"
    username = None
    password = None

    @staticmethod
    def set_config(host: str, port: int, database: str, username: str = None, password: str = None):
        """
        Thiết lập thông tin kết nối MongoDB (dùng thuộc tính tĩnh).
        """
        MongoDBConfig.host = host
        MongoDBConfig.port = port
        MongoDBConfig.database = database
        MongoDBConfig.username = username
        MongoDBConfig.password = password

    @staticmethod
    def get_connection_uri() -> str:
        """
        Tạo URI kết nối MongoDB từ thông tin tĩnh.
        :return: Chuỗi URI kết nối.
        """
        if MongoDBConfig.username and MongoDBConfig.password:
            return f"mongodb://{MongoDBConfig.username}:{MongoDBConfig.password}@{MongoDBConfig.host}:{MongoDBConfig.port}/{MongoDBConfig.database}"
        return f"mongodb://{MongoDBConfig.host}:{MongoDBConfig.port}/{MongoDBConfig.database}"
```

```python
def aggregate_data(connection_string, database_name, collection_name, pipeline):
    try:
        # Kết nối đến MongoDB
        client = MongoClient(connection_string)
        db = client[database_name]
        collection = db[collection_name]

        # Thực hiện aggregation
        result = collection.aggregate(pipeline)

        # Trả về kết quả dưới dạng danh sách
        return list(result)
    
    except Exception as e:
        print(f"An error occurred: {e}")
        return []
```

```python
def find_documents_by_field(collection, field_name, value):
    query = {field_name: value}
    return list(collection.find(query))
```

```python
class ComparisonConverter:
    def __init__(self):
        # Bảng ánh xạ giữa các phép so sánh và biểu thức MongoDB
        self.comparison_map = {
            '>': '$gt',
            '>=': '$gte',
            '<': '$lt',
            '<=': '$lte',
            '=': '$eq',
            '!=': '$ne',
            'in': '$in',
            'not in': '$nin',
        }

    def convertFormular(self, operator, value):

        # Kiểm tra phép so sánh có trong bảng ánh xạ không
        if operator not in self.comparison_map:
            raise ValueError(f"Operator '{operator}' is not supported.")
        elif(operator in ['>','>=','<','<=','=','!=']):
            if (is_number(value[0])==True):
                value = float(value[0])
                return(json.dumps({self.comparison_map[operator]: value}))
            else:
                if (operator in ['=','!=']):
                    return json.dumps({self.comparison_map[operator]: value})
                else:
                    raise ValueError("Please input a number.")
        else:
        # Trả về cú pháp MongoDB
            result = json.dumps({self.comparison_map[operator]: value})
            return result
    
    def convertDate(self,varDate,value):
        date = '"$'+ varDate + '"'
        value = "".join(map(str,value))
        Datediff = (
    '{"$expr": {'
    '"$lte": ['
    '{"$dateFromParts": {'
    '"year": { "$toInt": {"$substr": [' + date.lower() + ', 0, 4]} },'
    '"month": { "$toInt": {"$substr": [' + date.lower() + ', 4, 2]} },'
    '"day": { "$toInt": {"$substr": [' + date.lower() + ', 6, 2]} }'
    '}},'
    '{"$dateSubtract": {'
    '"startDate": start_date,'
    '"unit": "day",'
    '"amount": ' + str(value) +
    '}'
    '}'
    ']}}'
)
        return Datediff
```

```python
class Concat_Query:
    @staticmethod
    def concatOperator(lst_operator):
        query_cmd = ''
        for i in range(len(lst_operator)):       
            if lst_operator[i] != None: # điều kiện có operator
                operator = '{"$'+ lst_operator[i] +'"'
                query_cmd += operator.lower() + ': ['
        return query_cmd
    
    @staticmethod
    def concatGroup(lst_group):
        query_cmd = '{"$group": {"_id":'
        result = {item: "$" + str(item) for item in lst_group}
        query_cmd += json.dumps(result) + ","
        return query_cmd
```

```python
def init_Query_cmd(tbl_name,value):
    # tạo Table từ collection Config_query
    collection_name = "TM_CONFIG_QUERY"
    collection = MongoDBHandler(MongoDBConfig.get_connection_uri(),MongoDBConfig.database).database[collection_name]
    documents = find_documents_by_field(collection,tbl_name,value)
    df = pd.DataFrame(documents)
    table =[]
    for i in range(df['Table'].count()):
        table.append(df[['Table','Condition']].iloc[i])
        
    # Tạo câu lệnh aggregation cho Config_query    
    lst_query = {}
    lst_operator = []
    converter = ComparisonConverter()
    for i in range(len(table)):
        lst_operator.clear()
        if(table[i].Condition != None):
            for j in range(len(table[i].Condition)):
                lst_operator.append(table[i].Condition[j]['Operator'])
                if(table[i].Condition[j]['Formular'] == 'Datediff()' and lst_query.get(table[i].Table) == None): # hàm datediff array 0
                    cmd_convert = converter.convertDate(table[i].Condition[j]['Field_id'],table[i].Condition[j]['Value'])+','
                    lst_query.update({table[i].Table : cmd_convert })
                elif(table[i].Condition[j]['Formular'] == 'Datediff()' and lst_query.get(table[i].Table) != None ): # hàm datediff array n
                    cmd_convert = lst_query.get(table[i].Table) + converter.convertDate(table[i].Condition[j]['Field_id'],table[i].Condition[j]['Value']) + ']},'
                    lst_query.update({table[i].Table : cmd_convert })
                elif(lst_query.get(table[i].Table) != None ): # các hàm so sánh array n
                    cmd_convert = lst_query.get(table[i].Table) + '{"'+ table[i].Condition[j]['Field_id'] +'":'+ converter.convertFormular(table[i].Condition[j]['Formular'],table[i].Condition[j]['Value'])+'}'+']},'
                    lst_query.update({table[i].Table : cmd_convert })
                else: # các hàm so sánh array 0
                    cmd_convert = '{"'+ table[i].Condition[j]['Field_id'] +'":'+ converter.convertFormular(table[i].Condition[j]['Formular'],table[i].Condition[j]['Value'])+'},'
                    lst_query.update({table[i].Table : cmd_convert })
            strmatch = '{"$match":'
            Query_cmd = Concat_Query.concatOperator(lst_operator) + lst_query.get(table[i].Table)
            lst_query.update({table[i].Table : Query_cmd })
            if lst_query.get(table[i].Table):
                qry_cmd = strmatch + lst_query.get(table[i].Table)[:-1] +'},'
            else:
                qry_cmd    
            lst_query.update({table[i].Table : qry_cmd })
    return lst_query
```

```python
def update_document(database_name, collection_name, filter_condition, update_values):
    """
    Hàm cập nhật một document trong MongoDB.

    :param database_name: Tên cơ sở dữ liệu.
    :param collection_name: Tên collection.
    :param filter_condition: Điều kiện lọc (dictionary).
    :param update_values: Giá trị cập nhật (dictionary, sử dụng các toán tử như $set, $inc...).
    :return: Kết quả cập nhật (matched_count và modified_count).
    """
    try:
        # Kết nối đến MongoDB
        connection_string = "mongodb://127.0.0.1:27017/"
        client = MongoClient(connection_string)
        db = client[database_name]
        collection = db[collection_name]

        # Cập nhật document
        result = collection.update_one(filter_condition, update_values)

        # Trả về thông tin cập nhật
        return {
            "matched_count": result.matched_count,
            "modified_count": result.modified_count
        }
    except Exception as e:
        print(f"Lỗi: {e}")
        return None
```

```python
class MethodComputeConverter:
    def __init__(self):
        self.comparison_map = {
            'Total' : '$sum',
            'Count' : '$sum',
            'Max' : '$max',
            'Min' : '$min'
        }
        self.operations = {
        '+': '$add',
        '-': '$subtract',
        '*': '$multiply',
        '/': '$divide'
        }
        
    def convertMethodCompute(self, operator, value):

        # Kiểm tra phép so sánh có trong bảng ánh xạ không
        if operator not in self.comparison_map:
            raise ValueError(f"Operator '{operator}' is not supported.")
        elif (operator!='Count'):
        # Trả về cú pháp MongoDB
            result = f'"{operator}{value.capitalize()}"' + ":" + json.dumps({self.comparison_map[operator]: f"${value}"})
        else:
            result = f'"{operator}"' + ":" + json.dumps({self.comparison_map[operator]: 1})
        return result
    
    def generate_add_fields_json(self,new_field, existing_field, operation, value):

    
        if operation not in self.operations:
            raise ValueError("Phép toán không hợp lệ. Chọn một trong: +, -, *, /")

        # Tạo công thức $addFields
        add_fields_query = {
            "$addFields": {
                new_field: { self.operations[operation]: ["$" + existing_field, value] }
            }
        }

        # Chuyển đổi thành chuỗi JSON
        return json.dumps(add_fields_query)
```

```python
def init_Equation_cmd(tbl_name,value):
    collection_name = "TM_CONFIG_EQUATION"
    collection = MongoDBHandler(MongoDBConfig.get_connection_uri(),MongoDBConfig.database).database[collection_name]
    documents = find_documents_by_field(collection,tbl_name,value)
    df = pd.DataFrame(documents)
    cmd_query = ''
    converter = MethodComputeConverter()
    if(df['Compute_method'][0] == "Create a new variable"):
        cmd_query = converter.generate_add_fields_json(df['New_field_name'][0],df['Field_to_compute'][0],df['Field_formular'][0],df['Field_value'][0])+","
    elif(df['Group_by'][0]!=None):
        cmd_query += Concat_Query.concatGroup(df['Group_by'][0])
        cmd_query += converter.convertMethodCompute(df['Compute_method'][0], df['Field_id'][0])+ "}},"
        if (df['Formular'][0] != None):
            strmatch = '{"$match":'
            converter_Comparision = ComparisonConverter()
            variable = re.search(r'"(.*?)"', converter.convertMethodCompute(df['Compute_method'][0], df['Field_id'][0])).group(1)
            if(is_number_list(df['Value'][0])):
                cmd_convert = '{"'+ variable +'":' + converter_Comparision.convertFormular(df['Formular'][0],df['Value'][0]) + '}'
                cmd_query += strmatch + cmd_convert + '}'
            else: raise ValueError('Please input a number')
    return {value : cmd_query}
```

```python
def is_number(value):
    try:
        float(value)  # Thử chuyển đổi thành số thực
        return True
    except ValueError:
        return False
def is_number_list(lst):
    """Kiểm tra xem tất cả phần tử trong list có phải là số hay không."""
    return all(is_number(item) for item in lst)
```

```python
def generate_add_fields_json(new_field, existing_field, operation, value):
    """
    Hàm tạo chuỗi JSON cho công thức $addFields trong MongoDB.
    
    :param new_field: Tên field mới cần thêm
    :param existing_field: Tên field có sẵn để tính toán
    :param operation: Phép toán (add, subtract, multiply, divide)
    :param value: Giá trị số hoặc field để thực hiện phép toán
    :return: Chuỗi JSON của công thức $addFields
    """
    
    # Chọn phép toán phù hợp
    operations = {
        "+": "$add",
        "-": "$subtract",
        "*": "$multiply",
        "/": "$divide"
    }
    
    if operation not in operations:
        raise ValueError("Phép toán không hợp lệ. Chọn một trong: +, -, *, /")

    # Tạo công thức $addFields
    add_fields_query = {
        "$addFields": {
            new_field: { operations[operation]: ["$" + existing_field, value] }
        }
    }

    # Chuyển đổi thành chuỗi JSON
    return json.dumps(add_fields_query)
```

```python
def create_lookup_with_unwind(from_collection, let_vars, pipeline, as_field):
    """
    Tạo một $lookup stage với let, pipeline và kết hợp $unwind để làm phẳng kết quả.

    :param from_collection: Tên collection cần join.
    :param let_vars: Dictionary chứa các biến định nghĩa trong `let`.
    :param pipeline: Danh sách các pipeline stages áp dụng cho `from_collection`.
    :param as_field: Tên field mới để lưu kết quả join.
    :return: Danh sách các stages ($lookup và $unwind).
    """
    lookup_stage = {
        "$lookup": {
            "from": from_collection,
            "let": let_vars,
            "pipeline": pipeline,
            "as": as_field
        }
    }

    unwind_stage = {
        "$unwind": {
            "path": f"${as_field}",  # Đường dẫn đến mảng cần làm phẳng
            "preserveNullAndEmptyArrays": True  # Giữ lại các documents không có kết quả join
        }
    }

    return [lookup_stage, unwind_stage]
```

```python
def get_combined_info(combine_id):
    # Dữ liệu mẫu
    query_data = [
        {"Qry_id": "Q001", "Rule_id": "RL001", "Table": "Transaction"},
        {"Qry_id": "Q002", "Rule_id": "RL001", "Table": "Customer"}
    ]
    equation_data = [
        {"Qry_id": "Q001", "Eqt_id": "E001", "Eqt_cmd": "test cmd"},
        {"Qry_id": "Q002", "Eqt_id": "E002", "Eqt_cmd": "test cmd"}
    ]
    combine_data = [
        {"Combine_id": "C001", "Eqt_ids": ["E001", "E002"]}
    ]

    # Tìm Combine dựa trên combine_id
    combine_info = next((item for item in combine_data if item["Combine_id"] == combine_id), None)
    if not combine_info:
        return []

    result = []
    # Duyệt qua từng Eqt_id trong Combine
    for eqt_id in combine_info["Eqt_ids"]:
        # Tìm Equation dựa trên Eqt_id
        equation_info = next((item for item in equation_data if item["Eqt_id"] == eqt_id), None)
        if not equation_info:
            continue

        # Tìm Query dựa trên Qry_id từ Equation
        query_info = next((item for item in query_data if item["Qry_id"] == equation_info["Qry_id"]), None)
        if not query_info:
            continue

        # Tạo dictionary kết quả
        combined_dict = {
            "Eqt_id": eqt_id,
            "Table": query_info["Table"],
            "Eqt_cmd": equation_info["Eqt_cmd"],
            "Rule_id": query_info["Rule_id"]
        }
        result.append(combined_dict)

    return result


# Ví dụ sử dụng
combine_id = "C001"
output = get_combined_info(combine_id)
print(output)
```
