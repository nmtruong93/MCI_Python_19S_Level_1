```python
import random
```

```python
a = int(input('Số bí mật sẽ nằm trong khoảng từ:'))
b = int(input('đến:'))
sec_number = random.randint(a,b)
print(sec_number)
```

```python
for i in range(sec_number):
    x = int(input('Nhập vào số bạn dự đoán:'))
    if (x < sec_number): 
        print('Số này bé hơn số bí mật')
    elif (x > sec_number): 
        print('Số này lớn hơn số bí mật')
    else : break
print('Chúc mừng bạn đã đoán đúng!')
```

```python
a,b = [int(input('Số bí mật sẽ nằm trong khoảng từ:')),int(input('\t\t\tđến:'))]#a,b là khoảng số nguyên sẽ tạo ra số ngẫu nhiên bí mật
for i in range(b-1): # xử lý bắt lỗi trường hợp nhập a > b
    if(a > b): 
        print('Nhập sai khoảng, cần nhập lại!')
        a,b = [int(input('Số bí mật sẽ nằm trong khoảng từ:')),int(input('\t\t\tđến:'))]
    else : break
sec_number = random.randint(a,b) # tạo ngẫu nhiên số bí mật trong khoảng từ a tới b
count = 0
for count in range(b):
    x = int(input('Nhập vào số bạn dự đoán:'))
    count+=1
    if (x < sec_number): 
        print('Số này bé hơn số bí mật')
    elif (x > sec_number): 
        print('Số này lớn hơn số bí mật')
    else :
        break
if (count < b) : print('Chúc mừng bạn đã đoán đúng!')
else: print('Bạn đã đưa ra %d câu trả lời, bạn không thể đưa ra số lần trả lời bằng %d lần cho phép'%(count,b))
```
